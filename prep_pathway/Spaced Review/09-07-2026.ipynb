Given nums and integer k, choose exactly k non-empty subarrays (overlaps and repeats allowed). Value of a subarray = max - min. Maximize total value across all k chosen subarrays.
nums = [4,2,5,1], k = 3 → 12

```python
def sol(num, k):
    # as repeats are allowed, we can always choose the highest yield numbers. 
    # Is my unserstanding correct?
    min_yield = min(num)
    max_yield = max(num)
    return k * (max_yield - min_yield)

print(sol([4,2,5,1], 3))
```

Given patterns and word, count how many strings in patterns occur as a substring of word.
patterns = ["a","abc","bc","d"], word = "abc" → 3

```python
def sol(patterns, word):
    counter = 0
    for i in patterns:
        if i in word:
            counter += 1
    return counter

print(sol(["a","abc","bc","d"], "abc"))
print(sol(["a","a","a"], "ab"))
```

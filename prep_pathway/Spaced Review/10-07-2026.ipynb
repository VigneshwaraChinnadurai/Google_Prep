Given integer n, concatenate its non-zero digits in order to form x (if none, x = 0). Let sum = sum of x's digits. Return x * sum.
n = 10203004 → non-zero digits 1,2,3,4 → x = 1234, sum = 10 → 12340
n = 1000 → x = 1, sum = 1 → 1

```python
def concatenate(num):
    final_number_in_str = ''
    final_number = 0
    for i in str(num):
        if i != '0':
            # print("i",i)
            final_number_in_str += i
            final_number += int(i)
            # print(final_number)
            # print(final_number_in_str)
    if final_number_in_str == '' or final_number == 0:
        return 0
    else:
        return int(final_number_in_str) * final_number
print(concatenate(1000))
print(concatenate(10203004))
```

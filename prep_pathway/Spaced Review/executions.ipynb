```python
import sys

print("Python's own artificial cap:", sys.getrecursionlimit())

depth_reached = 0

def recurse(n):
    global depth_reached
    depth_reached = n
    recurse(n + 1)   # no base case on purpose - keep pushing frames

try:
    recurse(0)
except RecursionError as e:
    print(f"Crashed at depth {depth_reached}: {e}")
```

```python
def f():
    x = "f's own x"
    print("before calling g:", x)
    x = g()
    print("after g returns: ", x)   # did g() touching its own x change f's x?

def g():
    x = "g's own x"   # SAME variable name as f's
    print("inside g:        ", x)
    return x

f()
```

```python
def rec(n):
    if n == 0:
        return 1
    else:
        return 2 * rec(n-1)

rec(5)
```

```python
def rec(n):
    final = 1
    for _ in range(1, n+1):
        final *= 2
    return final

rec(5)
```

```python
def rec_recursive(n):
    if n == 0:
        return 1
    return 2 * rec_recursive(n - 1)

def rec_iterative(n):
    final = 1
    for _ in range(1, n + 1):
        final *= 2
    return final

N = 3000  # bigger than Python's default recursion ceiling (1000)

print("--- recursive version ---")
try:
    result = rec_recursive(N)
    print("succeeded, result has", result.bit_length(), "bits")
except RecursionError as e:
    print("CRASHED:", e)

print("--- iterative version ---")
try:
    result = rec_iterative(N)
    print("succeeded, result has", result.bit_length(), "bits")
except RecursionError as e:
    print("CRASHED:", e)
```

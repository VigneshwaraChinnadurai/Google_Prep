```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Presentation style
plt.rcParams['font.size'] = 14
plt.rcParams['font.family'] = 'sans-serif'

def create_slide(title, content_lines, slide_num, filename=None):
    """Create a presentation slide as a figure."""
    fig, ax = plt.subplots(figsize=(16, 9))
    ax.set_xlim(0, 16)
    ax.set_ylim(0, 9)
    ax.axis('off')
    
    # Background
    fig.patch.set_facecolor('#1a1a2e')
    ax.set_facecolor('#1a1a2e')
    
    # Header bar
    header = FancyBboxPatch((0.3, 7.2), 15.4, 1.5, boxstyle="round,pad=0.1",
                            facecolor='#16213e', edgecolor='#0f3460', linewidth=2)
    ax.add_patch(header)
    
    # Title
    ax.text(8, 8.0, title, ha='center', va='center', fontsize=22, 
            fontweight='bold', color='#e94560')
    
    # Content
    y_pos = 6.5
    for line in content_lines:
        if line.startswith('##'):
            ax.text(1, y_pos, line[2:], fontsize=16, fontweight='bold', color='#ffffff')
        elif line.startswith('•'):
            ax.text(1.5, y_pos, line, fontsize=13, color='#e0e0e0')
        else:
            ax.text(1, y_pos, line, fontsize=14, color='#ffffff')
        y_pos -= 0.7
    
    # Slide number
    ax.text(15.5, 0.3, f'{slide_num}', ha='right', fontsize=11, color='#666666')
    
    # Footer
    ax.text(0.5, 0.3, 'Vigneshwara Chinnadurai | MBA Project | Manipal University Jaipur',
            fontsize=9, color='#666666')
    
    plt.tight_layout()
    if filename:
        plt.savefig(f'../figures/{filename}', dpi=150, bbox_inches='tight', facecolor=fig.get_facecolor())
    plt.show()
    return fig

print("Slide generation functions ready.")
```

```python
# SLIDE 7: Key Finding - Small Data Advantage
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))
fig.patch.set_facecolor('#1a1a2e')

# Learning curves
train_sizes = [50, 100, 200, 400, 800]
hybrid_accs = [0.782, 0.831, 0.862, 0.887, 0.894]
classical_accs = [0.721, 0.803, 0.867, 0.899, 0.912]

ax1.set_facecolor('#16213e')
ax1.plot(train_sizes, hybrid_accs, 'o-', color='#e94560', linewidth=3, markersize=10, label='Hybrid Quantum')
ax1.plot(train_sizes, classical_accs, 's-', color='#4CAF50', linewidth=3, markersize=10, label='Classical NN')
ax1.axvspan(0, 150, alpha=0.1, color='#e94560')
ax1.set_xlabel('Training Samples', fontsize=13, color='white')
ax1.set_ylabel('Accuracy', fontsize=13, color='white')
ax1.set_title('Learning Curves: Quantum Advantage in Low-Data', fontsize=13, color='white')
ax1.legend(fontsize=11)
ax1.tick_params(colors='white')
ax1.grid(True, alpha=0.2)
ax1.text(75, 0.72, 'Quantum\nAdvantage\nZone', ha='center', fontsize=10, 
         color='#e94560', style='italic')

# Parameter efficiency
ax2.set_facecolor('#16213e')
models = ['Logistic\nReg', 'SVM', 'NN\n(small)', 'NN\n(large)', 'Quantum\nVQC']
params = [9, 50, 161, 945, 72]
accs = [0.845, 0.862, 0.891, 0.912, 0.889]

colors = ['#607D8B', '#607D8B', '#607D8B', '#607D8B', '#e94560']
ax2.scatter(params, accs, c=colors, s=200, zorder=5, edgecolors='white', linewidth=1.5)
for i, (p, a, m) in enumerate(zip(params, accs, models)):
    ax2.annotate(m, (p, a), textcoords="offset points", xytext=(0, 15), 
                ha='center', fontsize=9, color='white')
ax2.set_xlabel('Parameters', fontsize=13, color='white')
ax2.set_ylabel('Accuracy', fontsize=13, color='white')
ax2.set_title('Parameter Efficiency', fontsize=13, color='white')
ax2.tick_params(colors='white')
ax2.grid(True, alpha=0.2)

plt.suptitle('Key Finding: Quantum Advantage in Specific Regimes', 
             fontsize=16, color='#e94560', fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('../figures/slide_07_key_finding.png', dpi=150, bbox_inches='tight', facecolor=fig.get_facecolor())
plt.show()
```
